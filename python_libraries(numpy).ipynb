```python
#view v/s copy
#view means modifying the original data and copy means modifying the copied data which will not affect the original one
```

```python
 a = np.array([10,20,30,40,50,60,70,80])
 a
```

```python
b = a[1:4]
b[:] = 0
print(b)
print(a)
```

```python
#copy
a = np.array([10,20,30,40,50,60,70,80])
b = a[1:4].copy()
b[:] = 0
print(a)
```
