```python
#(1) .zeros() --> it will create an array in which all the elements will be 0 either one dimensional or multi- dimensional
```

```python
a = np.zeros(3)
a
```

```python
a = np.zeros((3,4))
a
```

```python
#(2) .ones --> It will create an array in which all the elements will be 1 either one-d or multi-d
```

```python
a = np.ones(3)
a
```

```python
a = np.ones((2,3))
a
```

```python
#(3) .eye() --> It will create an array in which diagonal positional elements will be 1 and rest all are 0
```

```python
a = np.eye(4)
a
```

```python
#(4) .diag() --> It will create an array in which we can set our custom value on diagonal position
```

```python
a = np.array([1,45,78,90])
a
```

```python
b = np.diag(a)
b
```
