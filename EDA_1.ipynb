```python
import pandas as pd
df = pd.read_csv("/content/titanic - titanic.csv")
```

```python
## 1. Univariate Analysis
```

```python
df.columns
```

```python
df["Survived"].value_counts()
```

```python
import seaborn as sns
sns.countplot(x = df["Survived"])
```

```python
# if we want to see actual values then we use value_counts()
df["Survived"].value_counts()
```

```python
df["Survived"].value_counts().plot(kind = "bar")
```

```python
# if we want to check for Pclass then
sns.countplot(x=df["Pclass"])
```

```python
df["Survived"].value_counts().plot(kind = "pie",autopct = "%.2f") ##autopct meand autopercentage and %.2f means till 2 float numbers
```
