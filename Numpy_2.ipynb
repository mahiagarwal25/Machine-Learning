```python
# 5a. Randint() --> Random + Integer
# It will create an array in which it will provide random numbers in a given range.
```

```python
# Syntax: np.random.randint(min_number, max_number, total)
```

```python
import numpy as np
a = np.random.randint(1,10,3)
a
```

```python
# 5b. Seed() --> It will fix your random generated data
```

```python
np.random.seed(78)
a = np.random.randint(1,10,3)
a
```
