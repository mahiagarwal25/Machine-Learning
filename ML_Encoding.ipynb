```python
# Encoding --> Transform your categorical data into numerical data
# 1.Label --> It will encode data 1-1 columns.
# 2.Ordinal --> It will encode data more than 1 columns.
# 3.OneHot --> It will create new column based on sub-categories but it has only values not column name.
# 4.get_dummies --> It will create new column based on sub-categories but it shows column name with values.
```

```python
#1. LabelEncoder
```

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("/content/toyota - toyota.csv")
df.head(3)
```

```python
df['model'].value_counts()
```

```python
x = df.drop(columns = ['price'])
y = df['price']
```

```python
x.head(3)
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
lb = LabelEncoder()
```

```python
x['model'] = lb.fit_transform(x['model'])
x['transmission'] = lb.fit_transform(x['transmission'])
x['fuelType'] = lb.fit_transform(x['fuelType'])
```

```python
x.head(3)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
np.round(x_train.describe(),2)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc = StandardScaler()
```

```python
x_train_sc = sc.fit_transform(x_train)
```

```python
x_train_new = pd.DataFrame(x_train_sc, columns = x_train.columns)
```

```python
np.round(x_train_new.describe(),2)
```

```python
# 2.Ordinal Encoder
```

```python
df = pd.read_csv("/content/covid_toy - covid_toy.csv")
```

```python
df = df.dropna()
```

```python
df = df.drop(columns = ['age','fever'])
```

```python
df.head(2)
```

```python
df['city'].value_counts()
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
oe = OrdinalEncoder(categories = [['Male','Female'],
                                 ['Mild','Strong'],
                                 ['Kolkata','Bangalore','Delhi','Mumbai'],
                                 ['Yes','No']])
```

```python
df_ar = oe.fit_transform(df)
```

```python
df_new = pd.DataFrame(df_ar, columns = df.columns)
```

```python
df_new
```

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("/content/covid_toy - covid_toy.csv")
df.head(3)
```

```python
df = df.dropna()
```

```python
df = df.drop(columns = ['age','fever'])
```

```python
df.head(2)
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder(drop = 'first',sparse_output = False, dtype = np.int32)
```

```python
# gender --> 2-1 =1
# cough --> 2-1 = 1
# city --> 4-1 = 3
# has_covid --> 2-1 = 1
```

```python
ohe.fit_transform(df[['gender','cough','city','has_covid']])
```

get_dummies

```python
a = pd.get_dummies(df, columns = ['gender','cough','city','has_covid'], drop_first = True)
```
