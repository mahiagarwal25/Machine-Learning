```python
import numpy as np
```

```python
# View vs Copy
# View means modification in original data.
# Copy means changes in duplicate data will not reflect original data
```

```python
a = np.array([10,20,30,40,50,60,70,80,90])
a
```

```python
b = a[3:6]
b[:] = 0
```

```python
a
```

```python
#Copy
a = np.array([10,20,30,40,50,60,70,80,90])
a
```

```python
b = a[3:6].copy()
b[:] = 0
```

```python
a
```

```python
#Unique() --> It will remove duplicate elements from an array
# a --> unique elements array
#return_index = True --> Return indexing of unique elements
#return_counts = True --> Total frequency of each unique item
```

```python
a = np.array([1,2,3,1,1,2,3,3,3,2,2,4])
a
```

```python
np.unique(a, return_index=True, return_counts=True)
```
